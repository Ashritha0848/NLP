```python


import spacy

# Load English model
nlp = spacy.load("en_core_web_sm")

print("Model loaded successfully!")
```

```python
sentences = [
    "The cat sits on the mat.",
    "She is reading a book.",
    "John and Mary went to the market.",
    "Although it was raining, we went outside.",
    "The project was completed by the team.",
    "Can you solve this problem?",
    "He quickly finished his homework.",
    "The beautiful garden attracted many visitors.",
    "She said that she would come.",
    "The car was repaired by the mechanic.",
    "AI is transforming the world.",
    "They are playing football in the field.",
    "If you study hard, you will succeed.",
    "The teacher gave students homework.",
    "Why are you late?"
]
```

```python
docs = [nlp(sentence) for sentence in sentences]
```

```python
import pandas as pd

data = []

for doc in docs:
    for token in doc:
        data.append([
            token.text,
            token.lemma_,
            token.pos_,
            token.dep_,
            token.head.text
        ])

df = pd.DataFrame(data, columns=["Token", "Lemma", "POS", "Dependency", "Head"])
print(df)
```

```python
for doc in docs:
    print("\nSentence:", doc.text)

    for token in doc:
        if token.dep_ == "nsubj":
            print("Subject:", token.text)
        if token.dep_ == "ROOT":
            print("Main Verb:", token.text)
        if token.dep_ == "dobj":
            print("Object:", token.text)
```

```python
from spacy import displacy

# Visualize first 5 sentences
for doc in docs[:5]:
    displacy.render(doc, style="dep", jupyter=True)
```
